# Joining an Additional Dataset

In this step, I join my original e-commerce data with a category lookup file (`Category_Details.csv`).

I use a **left join** on `Category` so I keep all original sales rows and only add matching category details where available.


```python
import pandas as pd 
import numpy as np 
```

```python
sales_df = pd.read_csv("../data/raw/ecommerce_sales_data.csv")
```

```python
cat_df = pd.read_csv("../data/raw/Category_Details.csv")
```

## Why left join?

I used a left join because my original e-commerce table is the main dataset for the project.
A left join keeps every original sales row, and adds columns from "Category_Details" only where the category matches.
If a category does not match, the new columns become NaN, which helps me spot unmatched categories. ezzzz 


```python
# Left join on Category
joined_df = sales_df.merge(cat_df, on="Category", how="left")

print("Joined shape:", joined_df.shape)
joined_df.head()
```
